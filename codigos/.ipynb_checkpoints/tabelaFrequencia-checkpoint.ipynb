```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
here = os.getcwd()
dataG = pd.read_csv(os.path.join(here,"..","dataset","graduate.csv"), sep=";")
dataD = pd.read_csv(os.path.join(here,"..","dataset","dropout.csv"), sep=";")
dataE = pd.read_csv(os.path.join(here,"..","dataset","enrolled.csv"), sep=";")
```

```python
columG = dataG.columns
columD = dataD.columns
columE = dataE.columns
```

```python
for n in range(len(columG)):
    contagem = dataG[columG[n]].value_counts().sort_index()
    plt.figure(figsize=(10,6))
    plt.bar(
        contagem.index.astype(str),
        contagem.values
    )
    plt.xlabel("valor")
    plt.ylabel("Quantidade")
    plt.title(columG[n])
    plt.savefig(os.path.join(here,"..","graficoFrequencia",f"{columG[n]}.png"),dpi=300)
    plt.close()
```
