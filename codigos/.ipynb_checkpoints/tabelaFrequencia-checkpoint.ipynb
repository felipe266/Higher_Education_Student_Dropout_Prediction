```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
here = os.getcwd()
dataSet = pd.read_csv(os.path.join(here,"..","dataset","data.csv"), sep=";")
```

```python
colum = dataSet.columns
```

```python
n = 36
#plt.figure(figsize=(10,6))
contagem = dataSet[colum[n]].value_counts()
contagem.plot(kind = "bar")
plt.xlabel("valor")
plt.ylabel("Quantidade")
plt.title(colum[n])
```
