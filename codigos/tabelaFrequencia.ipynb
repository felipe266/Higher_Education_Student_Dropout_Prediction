```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
here = os.getcwd()
dataSet = pd.read_csv(os.path.join(here,"..","dataset","data.csv"), sep=";")
```

```python
colum = dataSet.columns
```

```python
n = 2
#plt.figure(figsize=(10,6))
contagem = dataSet[colum[n]].value_counts()
plt.bar(
    contagem.index.astype(str),
    contagem.values
)
plt.xlabel("valor")
plt.ylabel("Quantidade")
plt.title(colum[n])
```
