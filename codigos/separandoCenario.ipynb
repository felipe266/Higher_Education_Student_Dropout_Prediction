```python
import pandas as pd
import numpy as np
import os
```

```python
here = os.getcwd()
dataSet = pd.read_csv(os.path.join(here,"..","dataset","data.csv"), sep=";")
```

```python
data = np.array(dataSet)
colunas = dataSet.columns.to_list()
```

```python
data[0]
```

```python
dropout = dataSet.loc[
                    dataSet["Target"] == "Dropout",
                    colunas[0:21]+colunas[33:36]
]
graduate = dataSet.loc[
                    dataSet["Target"] == "Graduate",
                    colunas[0:27]+colunas[33:36]
]
enrolled = dataSet.loc[
                    dataSet["Target"] == "Enrolled",
                    colunas[0:36]
]
```

```python
dropout.to_csv(os.path.join(here,"..","dataset","dropout.csv"), sep=";", index=False)
```

```python
graduate.to_csv(os.path.join(here,"..","dataset","graduate.csv"), sep=";", index=False)
```

```python
enrolled.to_csv(os.path.join(here,"..","dataset","enrolled.csv"), sep=";", index=False)
```
