```python
import pandas as pd
import numpy as np
import os
```

```python
here = os.getcwd()
dataSet = pd.read_csv(os.path.join(here,"..","dataset","data.csv"), sep=";")
```

```python
data = np.array(dataSet)
colunas = dataSet.columns.to_list()
```

```python
#loc[linha, coluna]
cenarioA = dataSet.loc[
                    :,
                    colunas[0:21]+colunas[33:37]
]
cenarioB = dataSet.loc[
                    :,
                    colunas[0:27]+colunas[33:37]
]
cenarioC = dataSet.loc[
                    :,
                    colunas[0:37]
]
```

```python
cenarioA.to_csv(os.path.join(here,"..","dataset","cenarioA.csv"), sep=";", index=False)
```
